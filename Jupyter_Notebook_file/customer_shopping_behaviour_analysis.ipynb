```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head(10)
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_category'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age_category', 'age']].head(20)
```

```python
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_day'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_day', 'frequency_of_purchases']]
```

```python
df[['discount_applied', 'promo_code_used']]
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
from sqlalchemy import create_engine

username = "postgres"
password = "Password"      # Your PostgreSQL password
host = "localhost"
port = "5432"
database = "postgres"      # Correct database

engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# Existing table name
table_name = "customer_shopping_analysis"

# Import the DataFrame
df.to_sql(
    name=table_name,
    con=engine,
    if_exists="append",    # Use "append" if the table already exists
    index=False
)

print(f"Data successfully imported into table '{table_name}' in database '{database}'.")
```
